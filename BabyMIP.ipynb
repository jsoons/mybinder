# Eenvoudig Optimalisatieprobleem

**Maximaliseer Doelfunctie:**

$$\max \quad 40x_A + 30x_B + 50x_C$$

**Constraintvoorwaarden:**

$$2x_A + 1x_B + 3x_C \leq 100$$

$$10y_i \leq x_i \leq 100y_i \quad \forall i \in \{A, B, C\}$$ (Big-M en minimaal 10 stuks per batch)

**Met:**

$$x_i \geq 0 \quad \forall i \in \{A, B, C\}$$

$$y_i \in \{0, 1\} \quad \forall i \in \{A, B, C\}$$

```python
from mip import *
m = Model()
x= [m.add_var(name='productiehoeveelheid') for i in range(3)]
y= [m.add_var(name='produceren' ,var_type=BINARY) for i in range(3)] #in probleem element van verzameling 0,1, nu binair, alternatief is INTEGER lb=0, ub=1

# add constraints
m+= 2*x[0]+1*x[1]+3*x[2]<=100

m+=10*y[0]<=x[0]<=100*y[0]

#m+= x >= 0

for i in range(3):
    m+= x[i] >=0


#[m.add_constr(xi >= 0) for xi in x]


#for i in range(3):
#    m+= x[i] >=0
    

# m += sum(x[:3] >= 0)

m.objective = maximize(40*x[0]+30*x[1]+50*x[2])


m.optimize()

```

```python
from mip import *

m = Model()
x = [m.add_var(name=f'productiehoeveelheid_{i}', var_type=INTEGER) for i in range(3)]
y = [m.add_var(name=f'produceren_{i}', var_type=BINARY) for i in range(3)]

# Add constraints
m += 2*x[0] + 1*x[1] + 3*x[2] <= 100

for i in range(3):
    m += x[i] >= 0
    m += 10*y[i] <= x[i]
    m += x[i] <= 100*y[i]

m+=x[1]<=10 #creatie van een tekort
m+=x[2]>=15 #verplichte productie


# Set objective to maximize
m.objective = maximize(40*x[0] + 30*x[1] + 50*x[2])

# Optimize
m.optimize()

# Print results
print(f"Status: {m.status}")
print(f"Objective value: {m.objective_value}")
for i in range(3):
    print(f"x[{i}] = {x[i].x}, y[{i}] = {y[i].x}")
```
